````sysml
package CarSample {
    private import ISQ::*;   // MassValue, PowerValue などをまとめてインポート
    private import SI::*;    // litre, kilowatt などの単位をインポート

    part def Engine {
        attribute displacement :> ISQSpaceTime::volume;  // 体積
        attribute maxPower     :> ISQ::power;            // 出力
    }
    part EngineA : Engine {                  // 型をEngineに設定
        :>> displacement = 2.0 [SI::litre];  // Engineの体積を再定義
        :>> maxPower = 206 [SI::kilowatt];   // Engineの出力を再定義
    }
    part EngineB : Engine {                  // 型をEngineに設定
        :>> displacement = 1.8 [SI::litre];  // Engineの体積を再定義
        :>> maxPower = 150 [SI::kilowatt];   // Engineの出力を再定義
    }
}
````

````sysml
%viz  --view Tree --style LR --style ortholine CarSample
````

````sysml
package L11_Attributes_Quantities {
    private import ScalarValues::Real;
    private import ISQ::length;
    private import ISQ::mass;
    private import SI::kg;
    private import SI::cm;

    attribute def ReindeerFeatures {
        // Not variables, so cannot be constant!
        attribute noseColor : Color;
        attribute eyeColor : Color;
        attribute antlerLength subsets ISQ::length;
    }
    
    part def Reindeer {
        constant attribute features : ReindeerFeatures;
        constant attribute weight redefines ISQ::mass default 110 [kg];
        attribute energyLevel : Real default := 100;
    }

    part def Rudolph specializes Reindeer {
        constant attribute :>> features = new ReindeerFeatures(Color::red, Color::brown, 60 [cm]);
        constant attribute :>> weight = 100 [kg];
        attribute :>> energyLevel := 200;
    }

    enum def Color {
        enum orange; enum yellow; enum green;
        enum blue;   enum indigo; enum violet;
        enum silver; enum gold;   enum red;
        enum brown;  enum white;  enum black;
    }
}
````

````sysml
%viz  --view Tree --style LR --style ortholine L11_Attributes_Quantities
````

````sysml
package AutomotiveSystem {
    private import ScalarValues::*;       // String を使うために追加
    private import ISQ::*;
    private import ISQMechanics::*;
    private import ISQElectromagnetism::*;
    private import ISQSpaceTime::*;
    private import SI::*;

    // =========================================================
    // attribute def
    // =========================================================

    attribute def FuelType {
        attribute name : String;          // ① ScalarValues::* のインポートで解決
    }

    // =========================================================
    // part def
    // =========================================================

    part def Engine {
        attribute displacement :> volume;
        attribute maxPower     :> power;
        port exhaustPort       : ExhaustPort;
    }

    // specializes :>
    part def ElectricMotor :> Engine {
        // ② ISQElectromagnetism::* をインポートして energy のサブセットで表現
        attribute batteryCapacity :> energy;

        // redefines :>>（行の途中）
        attribute zeroDisplacement :>> displacement = 0 [SI::litre];
    }

    part def Wheel {
        // subset :>
        attribute wheelDiameter :> length;
        attribute tirePressure  :> pressure;
    }

    part def Vehicle {
        attribute totalMass :> mass;
        attribute fuelType  : FuelType;

        part engine  : Engine;
        part wheels  : Wheel[4];

        port fuelPort : FuelPort;
    }

    // specializes :>
    part def ElectricVehicle :> Vehicle {
        // redefines :>>（行の途中）
        part motorEngine : ElectricMotor :>> engine;
    }

    // =========================================================
    // port def
    // =========================================================

    port def FuelPort {
        attribute flowRate :> volumeFlowRate;
    }

    port def ExhaustPort {
        attribute exhaustTemp :> temperature;
    }

    // =========================================================
    // トップレベルの part usage
    // =========================================================

    part familyCar : Vehicle {
        // redefines :>>（行の先頭）
        :>> totalMass = 1450 [SI::kg];

        part :>> engine : Engine {
            :>> displacement = 2.0  [SI::litre];
            :>> maxPower     = 115  [SI::kilowatt];
        }
        part :>> wheels : Wheel {
            :>> wheelDiameter = 430  [SI::mm];        // ③ 17インチ≒430mm
            :>> tirePressure  = 230000 [SI::Pa];      // ③ 230kPa = 230000Pa
        }
    }

    part electricSportsCar : ElectricVehicle {
        // redefines :>>（行の先頭）
        :>> totalMass = 2100 [SI::kg];

        // redefines :>>（行の途中）
        part sportMotor : ElectricMotor :>> motorEngine {
            :>> batteryCapacity = 360000000 [SI::J];  // ③ 100kWh = 3.6×10^8 J
            :>> maxPower        = 450 [SI::kilowatt];
        }
        part :>> wheels : Wheel {
            :>> wheelDiameter = 533  [SI::mm];        // ③ 21インチ≒533mm
            :>> tirePressure  = 250000 [SI::Pa];      // ③ 250kPa = 250000Pa
        }
    }
}
````

````sysml
%viz  --view Tree --style LR --style ortholine AutomotiveSystem
````

````sysml
package AutomotiveSystem2 {
    private import ScalarValues::*;
    private import ISQ::*;
    private import ISQMechanics::*;
    private import ISQElectromagnetism::*;
    private import ISQSpaceTime::*;
    private import SI::*;
    attribute def FuelType {
        attribute name : String;
    }
    part def Engine {
        attribute displacement :> volume;
        attribute maxPower :> power;
        port exhaustPort : ExhaustPort;
    }
    part def ElectricMotor :> Engine {
        attribute batteryCapacity :> energy;
        attribute zeroDisplacement :>> displacement = 0 [SI::litre];
    }
    part def Wheel {
        attribute wheelDiameter :> length;
        attribute tirePressure :> pressure;
    }
    part def Vehicle {
        attribute totalMass :> mass;
        attribute fuelType : FuelType;
        part engine : Engine;
        part wheels : Wheel [4];
        port fuelPort : FuelPort;
    }
    part def ElectricVehicle :> Vehicle {
        part motorEngine : ElectricMotor :>> engine;
    }
    port def FuelPort {
        attribute flowRate :> volumeFlowRate;
    }
    port def ExhaustPort {
        attribute exhaustTemp :> temperature;
    }
    part familyCar : Vehicle {
        :>> totalMass = 1450 [SI::kg];
        part :>> engine : Engine {
            :>> displacement = 2.0 [SI::litre];
            :>> maxPower = 115 [SI::kilowatt];
        }
        part :>> wheels : Wheel {
            :>> wheelDiameter = 430 [SI::mm];
            :>> tirePressure = 230000 [SI::Pa];
        }
    }
    part electricSportsCar : ElectricVehicle {
        :>> totalMass = 2100 [SI::kg];
        part sportMotor : ElectricMotor :>> motorEngine {
            :>> batteryCapacity = 360000000 [SI::J];
            :>> maxPower = 450 [SI::kilowatt];
        }
        part :>> wheels : Wheel {
            :>> wheelDiameter = 533 [SI::mm];
            :>> tirePressure = 250000 [SI::Pa];
        }
    }
    comment 
    /* part defは、ブループリント(設計書)そのもの。
     * 下記の記載は、「「Cockpitを持ったF1」を作るための設計書」と捉えるべき。
     *
     * ```
     * part def F1 :Car {
     *     part Cockpit: Chair;
     * }
     * ```
     * 下記の記載は、「「Cockpitの設計書を持ったF1」を作るための設計書」と捉えるべき。
     * F1に乗ろうとしたら、座るべき場所にはCockpitの設計書がポツンと置いてあるイメージ。
     * ```
     * part def F1 :Car {
     *     part def Cockpit: Chair;
     * }
     * ```*/
}
````

````sysml
%viz  --view Tree --style LR --style ortholine AutomotiveSystem2
````
